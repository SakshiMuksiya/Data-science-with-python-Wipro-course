```python
# 1 Write a function to return the sum of all numbers in a list.
# eg: (8, 2, 3, 0, 7) -> 20
def sum_of_list(list1):
    summ = 0
    for i in list1:
        summ += i
    return summ
list1 = []
n = int(input("Enter the number of elements you want to enter : "))
for i in range (1,n+1):
    k = int(input(f"Enter the {i} element: "))
    list1.append(k)
print(f"The sum of the given list is {sum_of_list(list1)}.")
```

```python
# 2 Write a function to return the reverse of a string
# eg: "123abcd" -> "dcba4321"
def reverse_of_string(string):
    return string[::-1]
string = input("Enter the string: ")
print(f"The reverse of given string is {reverse_of_string(string)}.")
```

```python
# 3 Write a function to calculate and return the factorial of a non-negative number.
def factorial(num):
    if num <= 0:
        return 1
    else:
        fact = 2
        for i in range(3,num+1):
            fact *= i
        return fact
num = int(input("Enter the number whose factorial you want to calculate: "))
print(f"The factorial of {num} is {factorial(num)}. ")
```

```python
# 4 Write a function that accepts a string and prints the number of uper case letters and lower case letters in it.
def count_upper_lower(string):
    count_upper = 0
    count_lower = 0
    for i in string:
        if i.isupper():
            count_upper += 1
        elif i.islower():
            count_lower += 1
    print("The number of upper case letters in given string: ",count_upper)
    print("The number of lower case letters in given string: ",count_lower)
string = input("Enter the string consists of upper case and lower case letters: ")
count_upper_lower(string)
```

```python
# 5 Write a function to print the even numbers from a given list. List passed as an argument.
# eg: [1, 2, 3, 4, 5, 6, 7, 8, 9] -> [2, 4, 6, 8]
def even_num(list2):
    dummy = []
    for i in list2:
        if i % 2 == 0:
            dummy.append(i)
    return dummy
list2 = []
leng = int(input("Enter the number of elements you want to enter: "))
for i in range(leng):
    k = int(input(f"Enter {i+1} element of list: "))
    list2.append(k)
print(f"The list of even numbers in the given list is {even_num(list2)}.")
```

```python
# 6 Write a function that takes a number as a parameter and checks whether the number is prime or not.
def is_prime(number):
    if number == 0 or number == 1:
        return False
    else:
        for i in range(2, int(number ** 0.5)+1):
            if number % i == 0:
                return False
        return True
number = int(input("Enter the number to check whether prime or not: "))
print(f"The number {number} is {"Prime" if is_prime(number) else "Not Prime"}.")
```

```python
# 1 Sort the colors
def sort_the_color(string):
    #sort the colors in alphabetical order
    colors = string.split('-')
    colors.sort()
    return '-'.join(colors)
colors = input("Enter the colors separated by '-' : ")
print(f"The sorted colors are {sort_the_color(colors)}.")
```
