```python
# 1 Write a program to check if a given number is Positive, Negative, or Zero.
num = float(input("Enter a number: "))
if num == 0:
    print("The number is Zero.")
elif num > 0:
    print(f"The number {num} is Positive.")
else:
    print(f"The number {num} is Negative.")
```

```python
# 2 Write a program to check if a given number is odd or even.
number = int(input("Enter a number: "))
if number % 2 == 0:
    print(f"The number {number} is Even.")
else:
    print(f"The number {number} is Odd.")
```

```python
# 3 Given two non-negative values, print true if they have the same last digit, such as 27 and 57.
# last_digit(7,17) → true
# last_digit(6,17) → false
def last_digit(num1, num2):
    return num1 % 10 == num2 % 10 if num1 >= 0 and num2 >= 0 else "Both numbers must be non-negative."

num1 = int(input("Enter the first non-negative number:"))
num2 = int(input("Enter the second non-negative number:"))
print(last_digit(num1, num2))
```

```python
# 4 Write a program to print numbers from 1 to 10 in a single row with one tab space.
for i in range (1, 11):
    print(i, end = "\t")
print()
```

```python
# 6 Write a program to check if a given number is prime or not.
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
prime_num = int(input("Enter a number to check if it's prime: "))
print(f"The number {prime_num} is {'Prime' if is_prime(prime_num) else 'Not Prime'}.")
```

```python
# 7 Write a program to print prime numbers between 10 and 99.
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
for num in range(10, 100):
    if is_prime(num):
        print(num)
```

```python
# 8 Write a program to print the sum of all the digits of a given number.
# eg: 1234 -> 10
sum_digit = int(input("Enter a number to calculate it's digit sum: "))
digit_sum = sum(int(digit) for digit in str(abs(sum_digit)))
print(f"The sum of the digits of {sum_digit} is {digit_sum}.")
```

```python
# 9 Write a program to reverse a given number and print.
# eg: 1234 -> 4321
rev_num = int(input("Enter a number to reverse: "))
print(f"The reverse of {rev_num} is {str(abs(rev_num))[::-1] if rev_num >= 0 else '-' + str(abs(rev_num))[::-1]}.")
```

```python
# 10 Write a program to find if the given number is palindrome or not.
# eg: 110011 -> Palindrome
p_num = int(input("Enter a number to check if it's a palindrome: "))
p_str = str(abs(p_num))
print(f"The number {p_num} is {"a Palindrome." if p_str == p_str[::-1] else "Not a Palindrome."}")
```
