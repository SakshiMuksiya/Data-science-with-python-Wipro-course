```python
import re
```

```python
#Write a program to find check if a string has only octal digits. Given string ['789', '123', '004']

strings = ['789', '123', '004']
for s in strings:
    if re.fullmatch(r'[0-7]+', s):
        print(f"{s} has only octal digits.")
    else:
        print(f"{s} does not have only octal digits.")
```

```python
#Extract the user id, domain name and suffix from the given email address. Given email address ['zuck@facebook.com', 'page@facebook.com', 'sunder33@google.com']

emails = ['zuck@facebook.com', 'page@facebook.com', 'sunder33@google.com']
for email in emails:
    match = re.match(r'([^@]+)@([^@]+)\.([^@]+)', email)
    if match:
        user_id = match.group(1)
        domain_name = match.group(2)
        suffix = match.group(3)
        print(f"Email: {email}")
        print(f"User ID: {user_id}")
        print(f"Domain Name: {domain_name}")
        print(f"Suffix: {suffix}")
        print()
```

```python
#Split the following irregular sentence into proper words
#sentence = """A, very      very;      irregular_sentence"""
#output = A very very irregular sentence
sentence = "A, very      very;      irregular_sentence"
words = re.findall(r'\b\w+\b', sentence)
print(" ".join(words))
```

```python
#Clean up the following tweet so that it contains only the users message i.e. remove all URLs, hashtags, mentions and reserved words
#tweet = '''#Good advice! RT @TheNextWeb: What I would do differently if I was learning to code today http://t.co/lbwej0pxOd cc: @garybernhart #rstats #programming'''
#output = Good advice! What I would do differently if I was learning to code today
tweet = '''Good advice! RT @TheNextWeb: What I would do differently if I was learning to code today http://t.co/lbwej0pxOd cc: @garybernhart #rstats #programming'''
cleaned_tweet = re.sub(r'http\S+|@\S+|#\S+|RT|cc:', '', tweet)
cleaned_tweet = re.sub(r'\s+', ' ', cleaned_tweet).strip()
print(cleaned_tweet)
```

```python
import requests
import re

url = "https://raw.githubusercontent.com/selva86/datasets/master/13/sample.html"

r = requests.get(url)
html = r.text

print(r.status_code)
print(r.text)
output = re.findall(r'>([^<>]+)<', html)

# Remove empty strings and whitespace
output = [text.strip() for text in output if text.strip()]

print(output)
```

```python
#Given below list of words, identify the words tha begin and end with the same character
'''
civic
trust'
widows
maximum
museums
aa
as
'''
words = ['civic', 'trust', 'widows', 'maximum', 'museums', 'aa', 'as']
same_char_words = [word for word in words if word[0] == word[-1]]
print("Words that begin and end with the same character:", same_char_words)
```
