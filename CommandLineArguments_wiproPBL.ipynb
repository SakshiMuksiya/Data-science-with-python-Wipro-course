```python
# 1 Write a program to accept two numbers as command line arguments and display their sum.
import sys

sys.argv = ['sum.py', '10', '20']

num1 = float(sys.argv[1])
num2 = float(sys.argv[2])

print("Sum =", num1 + num2)
```

```python
# 3 Write a program to accept 10 numbers through command line arguments and calculate the sum of prime numbers among them.

# Simulate 10 command-line arguments
sys.argv = ['prime_sum.py', '10', '7', '15', '11', '8', '13', '21', '17', '4', '19']

# Function to check if a number is prime
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_sum = 0

# Read the 10 numbers from sys.argv
for i in range(1, 11):
    num = int(sys.argv[i])
    if is_prime(num):
        prime_sum += num

print("Sum of prime numbers =", prime_sum)
```

```python
import sys

# Simulate command-line arguments
sys.argv = [
    "happiness.py",
    "1-3-5-7",      # String 1
    "2-4-6-8",      # String 2
    "1-2-3-4-5-9"   # String 3
]

# Split the strings into lists
string1 = sys.argv[1].split('-')
string2 = sys.argv[2].split('-')
string3 = sys.argv[3].split('-')

happiness = 0

# Calculate happiness
for num in string3:
    if num in string1:
        happiness += 1
    elif num in string2:
        happiness -= 1

print("String 1:", string1)
print("String 2:", string2)
print("String 3:", string3)
print("Happiness =", happiness)
```
