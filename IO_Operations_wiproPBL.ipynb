```python
import re
from collections import Counter

file_name = "apollo_story.txt"

# --- Part 1: Determine Meeting Time based on line count ---
line_count = 0
with open(file_name, "r") as f:
    for line in f:
        line_count += 1

meeting_time = ""
if 1 <= line_count <= 24:
    if line_count <= 12:
        meeting_time = f"{line_count} AM"
    else:
        meeting_time = f"{line_count - 12} PM"
    print(f"Meeting Time: {meeting_time}")
else:
    print("Invalid number of lines for meeting time (must be between 1 and 24).")

# --- Part 2: Find Meeting Place based on most frequent word ---
all_words = []
with open(file_name, "r") as f:
    content = f.read().lower() # Read all content and convert to lowercase
    # Use regex to find all words (alphanumeric sequences)
    raw_words = re.findall(r'\b\w+\b', content)

# Define common stop words to filter out for a more meaningful result
stop_words = {'the', 'a', 'an', 'is', 'was', 'in', 'and', 'to', 'of', 'with', 'she', 'he', 'her', 'had', 'that', 'by', 'for', 'when', 'at', 'from', 'it', 'but', 'on', 'as', 'out', 'this', 'we', 'need', 'be', 'will', 'are', 'if', 'not', 'or', 'so'}

# Filter out stop words
meaningful_words = [word for word in raw_words if word not in stop_words]

if meaningful_words:
    word_counts = Counter(meaningful_words)
    most_common_word = word_counts.most_common(1)[0][0]
    meeting_place = f"{most_common_word.capitalize()} Street"
    print(f"Meeting Place: {meeting_place}")
else:
    print("No meaningful words found in the file to determine a meeting place.")
```
