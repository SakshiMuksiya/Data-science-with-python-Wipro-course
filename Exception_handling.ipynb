```python
# 1 Write a program to accept two numbers from user and perform division
# If any exception occurs, print an error message or else print the result.
a,b = input("Enter two numbers separated by space: ").split()
try:
    a = float(a)
    b = float(b)
    result = a / b
    print("Result:", result)
except ValueError:
    print("Error: Please enter valid numbers.")
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
```

```python
# 2 Write a program to accept a number from the user and check whether it's prime or not.
# If user enters anything other than number, handle the exception and print an error message.
num = input("Enter a number: ")
try:
    num = int(num)
    if num < 2:
        print(f"{num} is not a prime number.")
    else:
        for i in range(2, int(num**0.5) + 1):
            if num % i == 0:
                print(f"{num} is not a prime number.")
                break
        else:
            print(f"{num} is a prime number.")
finally:
    print("Execution completed.")
```

```python
# 3 Write a program to accept the file name to be opened from the user
# If file exist print the contents of the file in title case or else handle the exception and print an error message.
file_name = input("Enter the file name to be opened: ")
try:
    with open(file_name, 'r') as file:
        content = file.read()
        print(content.title())
except FileNotFoundError:
    print(f"Error: The file '{file_name}' does not exist.")
```

```python
# 4 Declare a list with 10 integers and ask the user to enter an index.
# Check whether the number in that index is positive or negative number.
# If any invalid index is entered, handle the exception and print an error message.
list_of_integers = [10, -5, 3, -1, 7, -2, 0, 4, -8, 6]
index = input("Enter an index (0-9): ")
try:
    index = int(index)
    if 0 <= index < len(list_of_integers):
        if list_of_integers[index] > 0:
            print(f"The number at index {index} is positive.")
        elif list_of_integers[index] < 0:
            print(f"The number at index {index} is negative.")
        else:
            print(f"The number at index {index} is zero.")
    else:
        print("Error: Index out of bounds.")
except ValueError:
    print("Error: Please enter a valid integer.")
```

```python
filename = input("Enter the file name: ")

total_amount = 0
item_count = 0
free_items = 0
discount = 0

with open(filename, "r") as f:
    for line in f:
        line = line.strip()

        # Skip empty lines
        if not line:
            continue

        # Read discount from the last line
        if line.lower().startswith("discount"):
            discount = float(line.split()[1])
            continue

        # Split item name and price
        parts = line.rsplit(" ", 1)

        if len(parts) != 2:
            continue

        item_name, price = parts

        # Check for free item
        if price.lower() == "free":
            free_items += 1
        else:
            total_amount += float(price)

        item_count += 1

final_amount = total_amount - discount

print("No of items purchased:", item_count)
print("No of free items:", free_items)
print("Amount to pay:", total_amount)
print("Discount given:", discount)
print("Final amount paid:", final_amount)
```
