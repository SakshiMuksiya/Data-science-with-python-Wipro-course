```python
# 1 Write a program to create a list of 5 integers and display the list items.
# Acess individual items through indexing.
my_list = []
for i in range(5):
    num = int(input(f"Enter integer {i+1}: "))
    my_list.append(num)
print("The list items are:")
for i in range(len(my_list)):
    print(f"Item {i}: {my_list[i]}")
```

```python
# 2 Write a program to append a new item to the end of the list.
new_item = int(input("Enter a new item to append to the list: "))
my_list.append(new_item)
print("Updated list items are:", my_list)
```

```python
# 3 Write a program to reverse the order of the items in the list.
new_list = list(reversed(my_list))
print("Reversed list items are:", new_list)
```

```python
# 4 Write a program to print the number of occurrences of a specified element in a list.
element = int(input("Enter the element to count its occurrences in the list: "))
count = my_list.count(element)
print(f"The element {element} occurs {count} times in the list.")
```

```python
# 5 Write a program to append the items of list1 to list2 in the front.
list1 = [1, 2, 3]
list2 = [4, 5, 6]
print("List1: ", list1)
print("List2: ", list2)
list2 = list1 + list2
print("List after appending list1 to the front of list2: ", list2)
```

```python
# 6 Write a program to insert a new item before the second element in an existing list.
new_item = int(input("Enter the new item to insert: "))
my_list.insert(1, new_item)
print("List after inserting the new item before the second element: ", my_list)
```

```python
# 7 Write a program to remove the item from a specified index in a list.
index = int(input("Enter the index of the item to remove: "))
if 0 <= index < len(my_list):
    removed_item = my_list.pop(index)
    print(f"Removed item: {removed_item}")
    print("List after removing the item: ", my_list)
else:
    print("Invalid index. Please enter a valid index.")
```

```python
# 8 Write a program to remove the first occurrence of a specified element from a list.
element = int(input("Enter the element to remove its first occurrence from the list: "))
if element in my_list:
    my_list.remove(element)
    print(f"List after removing the first occurrence of {element}: ", my_list)
else:
    print(f"The element {element} not in list {my_list}. ")
```

```python
# 1 Write a program to add a key and value to a dictionary.
# eg: {0:10, 1:20} -> {0:10, 1:20, 2:30}
dict1 = {0: 10, 1: 20}
key = int(input("Enter the key to add to the dictionary: "))
value = int(input("Enter the value to add to the dictionary: "))
dict1[key] = value
print("Dictionary after adding the new key-value pair: ", dict1)
```

```python
# 2 Write a program to concatenate the following dictionaries to create a new one.
# eg: d1= {1:10, 2:20}  d2= {3:30, 4:40}  d3= {5:50, 6:60} -> {1:10, 2:20, 3:30, 4:40, 5:50, 6:60}
d1 = {1: 10, 2: 20}
d2 = {3: 30, 4: 40}
d3 = {5: 50, 6: 60}
new_dict = {**d1, **d2, **d3}
print("New dictionary after concatenation: ", new_dict)
```

```python
# 3 Write a program to check if a given key already exists in a dictionary
key_to_check = int(input("Enter the key to check in the dictionary: "))
if key_to_check in dict1:
    print(f"The key {key_to_check} exists in the dictionary.")
else:
    print(f"The key {key_to_check} does not exist in the dictionary.")
```

```python
# 4 Write a program to iterate over a dictionary using for loop and print the key alone, values alone and both keys and values.
for key in dict1:
    print(f"Keys only : {key}")
print()
for key in dict1:
    print(f"Values only : {dict1[key]}")
print()
for key in dict1:
    print(f"Key: {key}, Value: {dict1[key]}")
```

```python
# 5 Write a program to prepare a dictionary where the keys are numbers between 1 and 15 (both included) and the values are square of the keys.
new_dict = {}
for i in range(0,16):
    new_dict[i] = i**2
print("The generated dictionary is :", new_dict)
```

```python
# 6 Write a program to sum all the values in a dictionary, considering the values will be of int type.
summ = 0
for key in new_dict:
    summ += new_dict[key]
print("The dictionary is : ", new_dict)
print("The sum of values is: ",summ)
```

```python
# 5 Write a program to replace last value of tuples in a list to 100.
# eg: [(10, 20, 40), (40, 50, 60), (70, 80, 90)] -> [(10, 20, 100), (40, 50, 100), (70, 80, 100)]
tuple2 = [(10, 20, 40), (40, 50, 60), (70, 80, 90)]
new_tuple2 = []
for tup in tuple2:
    new_tup = tup[:-1] + (100,)
    new_tuple2.append(new_tup)
print("The modified list of tuples is: ", new_tuple2)
```

```python
# 1 Write a program to count the number of upper and lower case letters in a string.
count_u = 0
count_l = 0
string = input("Enter the string for counting upper and lower case letters: ")
for i in string:
    if i.isupper():
        count_u += 1
    elif i.islower():
        count_l += 1
print("The number of upper case letters in given string = ", count_u)
print("The number of lower case letters in given string = ", count_l)
```

```python
# 2 Write a program that will check whether a given string is palindrome or not.
string = input("Enter a string to check it's palindrome or not: ")
print(f"The given string is {string} and its reverse is {string[::-1]}.")
print(f"Hence, the given string is {"a Palindrome" if string == string[::-1] else "Not a Palindrome"}.")
```

```python
# 3 Given a string, return a new string made of n copies of the first 2 chars of the original string where n is the lenght of the string.
# The string length will be >=2. If input is "Wipro", output should be "WiWiWiWiWi".
string = input("Enter any string to modify: ")
n = int(input("Enter the number of copies you want of the first two characters of the given string: "))
print(f"The modified string is : {n*(string[:2])}")
```

```python
# 4 Given a string, if the first or last character is 'x', return the string without those 'x' character,
# else return the string unchanged. If input is "xHix", then output is "Hi".
x_string = 'xHellox'
if x_string[0] == 'x' or x_string[-1] == 'x':
    new_string = x_string.strip('x')
    print(new_string)
```

```python
# 5 Given a string and an integer n, return a string made of n repetitions of the last n characters of the string.
# You may assume that n is between 0 and the length of string inclusive.
# eg: Wipro and 3 -> "propropro".
string = input("Enter any string to modify: ")
n = int(input("Enter the number of characters to be taken from the end of string: "))
print(f"The modified string is : {n*(string[-n:])}")
```

```python
# 1 Write a program to remove a given item from the set.
set1 = {1,2,4,6,'d', 'a'}
item = eval(input("Enter the item you want to remove from the set: "))
set1.discard(item)
print("The modified set is: ", set1)
```

```python
# 2 Write a program to create a intersection of sets.
set2 = {1, 2, 3, 4, 5}
set3 = {4, 5, 6, 7, 8}
intersection = set2.intersection(set3)
print("The intersection of the two sets is: ", intersection)
```

```python
# 3 Write a program to create a union of sets.
union = set2.union(set3)
print("The union of the two sets is: ", union)
```

```python
# 4 Write a program to find the maximum and minimum value in a set.
max_value = max(union)
min_value = min(union)
print("The maximum value in the set is: ", max_value)
print("The minimum value in the set is: ", min_value)
```
